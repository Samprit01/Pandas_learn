```python
import pandas as pd
```

```python

data = {
    'Student': ['Alice', 'Bob', 'Charlie', 'Diana', 'Ethan'],
    'Math': [85, 90, 78, 92, 65],
    'Science': [88, 76, 95, 89, 70],
    'English': [90, 85, 80, 95, 75]
}

df = pd.DataFrame(data)

print("--- Original Data ---")
print(df)
```

```python
# View the first few rows
print(df.head())

# Get summary information about columns and data types
print(df.info())

# Get statistical summary (mean, min, max, etc.)
print(df.describe())
```

```python
# Calculate the mean across columns (axis=1) for Math, Science, and English
df['Average'] = df[['Math', 'Science', 'English']].mean(axis=1)

print("--- Data with Average Scores ---")
print(df[['Student', 'Average']])
```

```python
# Filter rows where Average is greater than 85
top_students = df[df['Average'] > 85]

print("--- Top Students (Average > 85) ---")
print(top_students[['Student', 'Average']])
```

```python
print("--- Subject Class Averages ---")
print(df[['Math', 'Science', 'English']].mean())
```

```python
# Save to a new CSV file
df.to_csv('student_grades_analyzed.csv', index=False)
print("File saved successfully!")
```
